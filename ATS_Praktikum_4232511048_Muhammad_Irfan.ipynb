```python
#soal no 7 Pemanas mencapai target
# Menentukan suhu awal dan target
suhu_saat_ini = 60
target = 85

# Perulangan untuk menaikkan suhu
while suhu_saat_ini < target:
    suhu_saat_ini += 5

    print(f"Suhu saat ini: {suhu_saat_ini} C")

    # Memberikan peringatan saat suhu mencapai 80
    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

# Menampilkan pesan setelah target tercapai
print(f"Target {target} C tercapai.")
```

```python
#Soal 9 pembacaan vibrasi dengan sensor rusak
# Data vibrasi lima unit mesin
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

# Menghitung jumlah data tidak valid
jumlah_tidak_valid = 0

# Memeriksa setiap data
for n, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)

    except ValueError:
        print(f"Unit {n}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    # Peringatan jika vibrasi lebih dari 2.0
    if vibrasi > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")

    # Menghentikan pemeriksaan jika vibrasi lebih dari 3.0
    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

# Menampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

CARA KERJA SOAL NOMOR 9
1. Inisialisasi data: Program menyimpan lima data vibrasi mesin ke dalam list.
2. Pemeriksaan unit pertama: Nilai 1.2 berhasil dikonversi menjadi angka dan tidak menghasilkan peringatan.
3. Pemeriksaan unit kedua: Nilai 0.9 berhasil dikonversi dan tidak menghasilkan peringatan.
4. Pemeriksaan unit ketiga: Data ERR tidak dapat dikonversi menjadi angka sehingga program menampilkan pesan data tidak valid dan menambah penghitung menjadi 1.
5. Pemeriksaan unit keempat: Nilai 3.1 melebihi 2.0 sehingga muncul peringatan. Karena juga melebihi 3.0, program menampilkan pesan kritis dan menghentikan perulangan menggunakan break.
6. Hasil akhir: Program menampilkan jumlah data tidak valid sebanyak 1.

```python
# PROGRAM PEMANTAUAN BEBAN LISTRIK HARIAN
# Masalah: Sebuah fasilitas mencatat beban listrik (kW) tiap jam.
# Program membaca data beban, mendeteksi anomali (beban terlalu tinggi / sensor rusak),
# lalu menghitung statistik dan estimasi konsumsi energi harian.

# Memuat
# 1. Data beban disimpan dalam list (bertipe teks, seperti hasil pembacaan sensor).
# 2. Perulangan for memeriksa data per jam.
# 3. try-except menangani data rusak (misal "N/A").
# 4. if-elif-else mengklasifikasikan status beban.
# 5. Setelah loop, hitung total energi (kWh), rata-rata, dan beban puncak.

# Batas beban (kW)
BATAS_NORMAL = 400
BATAS_WARNING = 500

# Data beban per jam (teks, karena dibaca dari sensor)
data_beban = ["320", "350", "410", "N/A", "520", "380", "-15", "450"]

beban_valid = []
jumlah_rusak = 0

print("=== PEMANTAUAN BEBAN LISTRIK HARIAN ===")

for jam, item in enumerate(data_beban, start=1):
    # Ubah teks menjadi angka; tangani data rusak
    try:
        beban = float(item)
        if beban < 0:
            raise ValueError("beban negatif")
    except ValueError:
        print(f"Jam {jam}: data tidak valid ({item}) - dilewati")
        jumlah_rusak += 1
        continue

    # Klasifikasi status beban
    if beban > BATAS_WARNING:
        status = "KRITIS"
    elif beban > BATAS_NORMAL:
        status = "WARNING"
    else:
        status = "NORMAL"

    beban_valid.append(beban)
    print(f"Jam {jam}: {beban:.1f} kW -> {status}")

# Ringkasan hasil (hanya jika ada data valid)
if len(beban_valid) > 0:
    energi_total = sum(beban_valid) * 1   # tiap data mewakili 1 jam
    rata_rata = energi_total / len(beban_valid)
    beban_puncak = max(beban_valid)

    print("\n=== RINGKASAN ===")
    print(f"Data valid       : {len(beban_valid)}")
    print(f"Data rusak       : {jumlah_rusak}")
    print(f"Energi total     : {energi_total:.2f} kWh")
    print(f"Beban rata-rata  : {rata_rata:.2f} kW")
    print(f"Beban puncak     : {beban_puncak:.2f} kW")
else:
    print("Tidak ada data valid.")
```

Cara kerja program untuk soal bonus:
1. Kita perlu menetapkan batas beban (400 dan 500 kW), siapkan data_beban (teks dari sensor), list kosong beban_valid, dan jumlah_rusak = 0.
2. Pada perulangan for: periksa data satu per satu dengan enumerate, nomor jam mulai dari 1.
3. Validasi try-except: ubah teks ke angka dengan float.
"N/A" gagal, ValueError muncul otomatis.
"-15" berhasil, tetapi negatif, jadi raise ValueError dipicu sendiri.
4. Tangani data rusak: cetak pesan, tambah jumlah_rusak, lalu continue untuk lanjut ke jam berikutnya.
5. Klasifikasi if-elif-else: di atas 500 = KRITIS, di atas 400 = WARNING, selain itu NORMAL.
6. Simpan dan cetak: masukkan beban ke beban_valid dengan append, lalu tampilkan hasilnya.
7. Ringkasan setelah loop: cek dulu ada data valid, lalu hitung total energi (sum), rata-rata (total dibagi len), dan puncak (max).
